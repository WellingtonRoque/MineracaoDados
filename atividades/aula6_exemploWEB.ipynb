```python
import requests

url = "https://www.weg.net/catalog/weg/BR/pt/Motores-El%C3%A9tricos/Trifásico---Baixa-Tensão/Uso-Geral/W22/W22-/W22-IR3-Premium-7-5-cv-4P-L112M-3F-220-380-V-60-Hz-IC411---TFVE---B35D/p/12460578"

resposta = requests.get(url)

print("Status:", resposta.status_code)
print("Tamanho:", len(resposta.text))
```

```python
import requests

url = "https://loja.herculesmotores.com.br/motor-eletrico-trifasico-standard/ip55/3cv/4p/220/380/motor-eltrico-trifsico-standard-ip55-3cv-4p-220-380-440v-b35d.html"

resposta = requests.get(url)

print("Status:", resposta.status_code)
print("Tamanho:", len(resposta.text))
```

1. Interpretar o HTML com BeautifulSoup

Crie uma nova célula e execute:

```python
from bs4 import BeautifulSoup

soup = BeautifulSoup(resposta.text, "html.parser")

print("Título da página:")
print(soup.title.text)
```

2. Vamos verificar o conteúdo da página

Depois execute:

```python
print(soup.get_text(" ", strip=True)[:5000])
```

3. Localizar as tabelas da página

Como estamos procurando especificações técnicas, vamos verificar se a página possui tabelas HTML:

```python
tabelas = soup.find_all("table")

print("Quantidade de tabelas encontradas:", len(tabelas))
```

Se aparecer, por exemplo:

Quantidade de tabelas encontradas: 3

podemos inspecionar cada uma:

```python
for i, tabela in enumerate(tabelas):
    print(f"\n--- TABELA {i} ---")
    print(tabela.get_text(" ", strip=True)[:2000])
```

4. Ver a estrutura das linhas da tabela

Execute esta célula:

```python
for i, tabela in enumerate(tabelas):

    print(f"\n========== TABELA {i} ==========")

    linhas = tabela.find_all("tr")

    print("Quantidade de linhas:", len(linhas))

    for linha in linhas:
        colunas = linha.find_all(["th", "td"])

        valores = [coluna.get_text(" ", strip=True) for coluna in colunas]

        print(valores)
```

5. Criar um dicionário com os dados extraídos

Vamos usar a primeira tabela:

```python
tabela = tabelas[0]

dados_motor = {}

linhas = tabela.find_all("tr")

for linha in linhas:
    colunas = linha.find_all(["th", "td"])

    valores = [coluna.get_text(" ", strip=True) for coluna in colunas]

    if len(valores) == 2:
        chave = valores[0]
        valor = valores[1]

        dados_motor[chave] = valor

dados_motor
```

6. Transformar em DataFrame

Agora vamos usar o Pandas:

```python
import pandas as pd

df_motor = pd.DataFrame([dados_motor])

df_motor
```

7. Selecionar somente os dados importantes

Para a nossa aula, não precisamos trabalhar inicialmente com todas as 24 informações. Vamos selecionar os principais dados técnicos do motor.

Execute:

```python
colunas_interesse = [
    "Descrição",
    "Potência",
    "Tensão",
    "Polaridade",
    "Frequência",
    "Rotação",
    "Corrente Nominal (A)",
    "Classe de Isolação",
    "Fator de Serviço",
    "Grau de Proteção",
    "Tipo de Carcaça",
    "Forma Construtiva",
    "Peso aprox."
]

df_tecnico = df_motor[colunas_interesse].copy()

df_tecnico
```

8. Verificar os tipos dos dados

Agora vem uma etapa muito importante de ETL.

Execute:

```python
df_tecnico.dtypes
```

Isso é justamente o problema de ETL que queremos demonstrar.

Agora execute a limpeza

Vamos fazer tudo em uma única célula, para evitar problemas de execução fora de ordem:

```python
# Limpeza e conversão dos dados

df_tecnico["Potência"] = (
    df_tecnico["Potência"]
    .astype(str)
    .str.replace("CV", "", regex=False)
    .str.strip()
    .astype(float)
)

df_tecnico["Frequência"] = (
    df_tecnico["Frequência"]
    .astype(str)
    .str.replace("Hz", "", regex=False)
    .str.strip()
    .astype(float)
)

df_tecnico["Rotação"] = (
    df_tecnico["Rotação"]
    .astype(str)
    .str.replace("RPM", "", regex=False)
    .str.strip()
    .astype(int)
)

df_tecnico["Fator de Serviço"] = (
    df_tecnico["Fator de Serviço"]
    .astype(str)
    .str.replace(",", ".", regex=False)
    .str.strip()
    .astype(float)
)

df_tecnico["Peso aprox."] = (
    df_tecnico["Peso aprox."]
    .astype(str)
    .str.replace("Kg", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
    .astype(float)
)

df_tecnico
```

8. Transformar a Tensão

Atualmente temos:

220/380/440V

Vamos retirar o V e separar os três valores:

```python
df_tecnico["Tensão"] = (
    df_tecnico["Tensão"]
    .astype(str)
    .str.replace("V", "", regex=False)
    .str.strip()
)

df_tecnico["Tensão"]
```

Agora podemos separar:

```python
tensoes = df_tecnico["Tensão"].str.split("/", expand=True)

tensoes.columns = ["Tensão_1", "Tensão_2", "Tensão_3"]

tensoes
```

E podemos transformar em números:

```python
tensoes = tensoes.astype(float)

tensoes
```

9. Fazer o mesmo com a corrente

A corrente veio assim:

8,89/5,13/4,44 A

```python
df_tecnico["Corrente Nominal (A)"] = (
    df_tecnico["Corrente Nominal (A)"]
    .astype(str)
    .str.replace("A", "", regex=False)
    .str.replace(",", ".", regex=False)
    .str.strip()
)
```

Agora:

```python
correntes = df_tecnico["Corrente Nominal (A)"].str.split("/", expand=True)

correntes.columns = [
    "Corrente_1",
    "Corrente_2",
    "Corrente_3"
]

correntes = correntes.astype(float)

correntes
```

10. Juntar tudo no DataFrame

Agora podemos incorporar essas novas colunas:

```python
df_tecnico = pd.concat(
    [
        df_tecnico,
        tensoes,
        correntes
    ],
    axis=1
)

df_tecnico
```

11. Conferir todas as colunas

Execute:

```python
print(df_tecnico.columns.tolist())

print(df_tecnico.shape)
```

12. Visualizar somente os dados numéricos

Agora vamos separar os atributos que podem ser utilizados diretamente em cálculos:

```python
colunas_numericas = [
    "Potência",
    "Frequência",
    "Rotação",
    "Fator de Serviço",
    "Peso aprox.",
    "Tensão_1",
    "Tensão_2",
    "Tensão_3",
    "Corrente_1",
    "Corrente_2",
    "Corrente_3"
]

df_numerico = df_tecnico[colunas_numericas]

df_numerico
```

Agora temos efetivamente dados prontos para análise matemática.
