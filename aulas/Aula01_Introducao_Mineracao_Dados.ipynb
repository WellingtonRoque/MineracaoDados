## 🔎 2. Quais dados podemos coletar?

Um motor pode ser monitorado por diferentes sensores e variáveis:

| Variável | Unidade | Possível utilização |
|---|---:|---|
| `corrente` | A | Identificar sobrecarga ou esforço |
| `tensao` | V | Verificar comportamento da alimentação |
| `vibracao` | mm/s | Identificar anomalias mecânicas |
| `temperatura` | °C | Identificar aquecimento |
| `rpm` | RPM | Acompanhar a rotação |
| `potencia` | kW | Avaliar consumo |
| `horas_operacao` | h | Relacionar comportamento com desgaste |
| `status` | — | Indicar condição do motor |

Um registro poderia ser representado assim:

```text
Motor M001
Temperatura = 76,4 °C
Vibração = 5,8 mm/s
Corrente = 15,2 A
RPM = 1728
Status = Falha
```

Mas um único registro diz pouco.

Imagine milhares ou milhões de registros.

É nesse cenário que a Mineração de Dados se torna importante.


## 💡 3. Dados, Informação e Conhecimento

### Dados

São registros brutos.

Exemplo:

`temperatura = 76,4`

### Informação

É o resultado da organização e interpretação dos dados.

Exemplo:

> A temperatura do motor aumentou significativamente.

### Conhecimento

É uma conclusão obtida a partir da análise de informações e padrões.

Exemplo:

> O aumento simultâneo de temperatura e vibração pode estar relacionado a situações de falha.

### Fluxo

```text
DADOS
  ↓
INFORMAÇÃO
  ↓
CONHECIMENTO
  ↓
DECISÃO
```

A Mineração de Dados participa desse processo buscando **padrões, relações e comportamentos relevantes** nos dados.


## ⛏️ 4. O que é Mineração de Dados?

**Mineração de Dados (Data Mining)** é o processo de explorar conjuntos de dados para descobrir padrões, relações, tendências e informações relevantes que podem apoiar a tomada de decisões.

Algumas perguntas que podemos fazer:

- Existe relação entre temperatura e vibração?
- O aumento da corrente acontece antes das falhas?
- Existem grupos de motores com comportamentos semelhantes?
- Podemos identificar automaticamente situações anormais?
- Podemos prever uma possível falha?
- Quais características são mais importantes para explicar uma falha?

Ao longo da disciplina aprenderemos técnicas para responder perguntas desse tipo.


## 🤖 5. Mineração de Dados, Data Science, Machine Learning e IA

Esses conceitos estão relacionados, mas não são sinônimos.

| Conceito | Ideia principal |
|---|---|
| **Data Science** | Utilizar dados, estatística, programação e conhecimento do negócio para gerar valor |
| **Mineração de Dados** | Descobrir padrões e relações relevantes nos dados |
| **Machine Learning** | Criar modelos capazes de aprender padrões a partir de dados |
| **Inteligência Artificial** | Desenvolver sistemas capazes de executar tarefas associadas à inteligência |
| **Business Intelligence** | Transformar dados em informações para apoiar decisões |

Durante a disciplina veremos como essas áreas podem trabalhar juntas.
