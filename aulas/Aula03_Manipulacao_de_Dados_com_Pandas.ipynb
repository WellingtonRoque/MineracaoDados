# 🔎 10. Filtrando dados

Uma das operações mais importantes em análise de dados é o filtro.

### Exemplo

Queremos visualizar somente os registros do motor M001.


````python
df[df["motor"] == "M001"]
````

Também podemos utilizar outras condições.

### Motores com temperatura acima de 70 °C


````python
df[df["temperatura"] > 70]
````

### Registros em situação de falha


````python
df[df["status"] == "Falha"]
````

### Combinação de condições

Podemos combinar condições utilizando `&`.

Exemplo:

> Temperatura acima de 70 °C **e** vibração acima de 3 mm/s.


````python
df[(df["temperatura"] > 70) & (df["vibracao"] > 3)]
````

# ↕️ 11. Ordenando os dados

Podemos ordenar uma coluna utilizando `sort_values()`.

### Maior temperatura primeiro


````python
df.sort_values("temperatura", ascending=False)
````

### Maior vibração primeiro


````python
df.sort_values("vibracao", ascending=False)
````

# ➕ 12. Criando novas colunas

Podemos criar novas informações a partir dos dados existentes.

Por exemplo, podemos calcular uma estimativa simples de energia consumida:

```text
energia ≈ potência × tempo
```

Para nosso exemplo, vamos criar uma coluna chamada `energia_estimada`.


````python
df["energia_estimada"] = df["potencia"] * 1

df[["motor", "potencia", "energia_estimada"]]
````

Também podemos criar uma classificação simples de temperatura.

> **Atenção:** os limites abaixo são apenas didáticos e não representam uma norma técnica para motores reais.


````python
def classificar_temperatura(temperatura):
    if temperatura < 65:
        return "Normal"
    elif temperatura < 75:
        return "Atenção"
    else:
        return "Alta"

df["nivel_temperatura"] = df["temperatura"].apply(classificar_temperatura)

df[["temperatura", "nivel_temperatura"]]
````

# 📊 13. Estatísticas básicas

O Pandas permite calcular rapidamente:

- média;
- mediana;
- mínimo;
- máximo;
- soma;
- desvio padrão.

Vamos começar pela temperatura.


````python
print("Média:", df["temperatura"].mean())
print("Mediana:", df["temperatura"].median())
print("Mínima:", df["temperatura"].min())
print("Máxima:", df["temperatura"].max())
````

Podemos também analisar várias variáveis de uma vez.


````python
df[["corrente", "vibracao", "temperatura", "rpm", "potencia"]].describe()
````

# 👥 14. Agrupando dados

Agora queremos responder:

> Qual é a temperatura média de cada motor?

Utilizaremos `groupby()`.


````python
df.groupby("motor")["temperatura"].mean()
````

Podemos agrupar e analisar várias variáveis.


````python
df.groupby("motor")[["temperatura", "vibracao", "corrente"]].mean()
````

Também podemos contar quantos registros existem em cada situação por motor.


````python
df.groupby(["motor", "status"]).size()
````
