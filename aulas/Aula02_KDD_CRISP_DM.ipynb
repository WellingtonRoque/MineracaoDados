# 🔎 1. Por que precisamos de um processo?

Na aula anterior começamos com uma pergunta:

Agora imagine que uma empresa entregue milhões de registros para você e diga:

Um projeto de Mineração de Dados precisa de um **processo organizado**.

Sem um processo, podemos:

Nesta aula conheceremos dois processos importantes:

- **KDD — Knowledge Discovery in Databases**
- **CRISP-DM — Cross Industry Standard Process for Data Mining**


# 🧠 2. KDD — Knowledge Discovery in Databases

KDD significa **Knowledge Discovery in Databases**, ou **Descoberta de Conhecimento em Bancos de Dados**.

O KDD descreve um processo para transformar dados em conhecimento.

Uma representação simplificada é:

```text
SELEÇÃO
   ↓
PRÉ-PROCESSAMENTO
   ↓
TRANSFORMAÇÃO
   ↓
MINERAÇÃO DE DADOS
   ↓
INTERPRETAÇÃO / AVALIAÇÃO
   ↓
CONHECIMENTO
```

Observe que **Mineração de Dados é uma etapa do processo KDD**, e não o processo inteiro.

Essa diferença é importante.


## 🔹 Etapa 1 — Seleção

Precisamos escolher quais dados serão utilizados.

No nosso projeto:

- Qual motor será analisado?
- Qual período?
- Quais sensores?
- Quais registros são relevantes?

Exemplo:

> Vamos analisar os dados dos motores M001 a M010 registrados durante os últimos 12 meses.

---

## 🔹 Etapa 2 — Pré-processamento

Os dados precisam ser verificados e tratados.

Podemos encontrar:

- valores ausentes;
- registros duplicados;
- valores inválidos;
- erros de medição;
- unidades diferentes;
- inconsistências.

---

## 🔹 Etapa 3 — Transformação

Os dados podem precisar ser transformados para facilitar a análise.

Exemplos:

- normalização;
- criação de novas variáveis;
- agregações;
- redução de dimensionalidade;
- conversão de formatos.

---

## 🔹 Etapa 4 — Mineração de Dados

Agora aplicamos técnicas para encontrar padrões.

Exemplos:

- classificação;
- agrupamento;
- associação;
- detecção de anomalias;
- modelos preditivos.

---

## 🔹 Etapa 5 — Interpretação e avaliação

Os resultados precisam fazer sentido.

Não basta o algoritmo produzir um resultado.

Precisamos perguntar:

> **Esse resultado é útil para o problema que estamos tentando resolver?**

Finalmente transformamos os resultados em conhecimento.


# 🔄 3. CRISP-DM

CRISP-DM significa:

**Cross Industry Standard Process for Data Mining**

É uma metodologia muito utilizada para organizar projetos de Mineração de Dados e Ciência de Dados.

Suas principais fases são:

```text
┌──────────────────────────────┐
│ 1. Entendimento do Negócio   │
└──────────────┬───────────────┘
               ↓
┌──────────────────────────────┐
│ 2. Entendimento dos Dados    │
└──────────────┬───────────────┘
               ↓
┌──────────────────────────────┐
│ 3. Preparação dos Dados      │
└──────────────┬───────────────┘
               ↓
┌──────────────────────────────┐
│ 4. Modelagem                 │
└──────────────┬───────────────┘
               ↓
┌──────────────────────────────┐
│ 5. Avaliação                 │
└──────────────┬───────────────┘
               ↓
┌──────────────────────────────┐
│ 6. Implantação               │
└──────────────────────────────┘
```

Mas existe uma característica muito importante:

> **O processo é iterativo.**

Podemos voltar para uma etapa anterior quando descobrimos um problema.


## 🔹 1. Entendimento do negócio

Antes de analisar os dados precisamos entender o problema.

No nosso caso:

> A indústria quer reduzir paradas inesperadas causadas por falhas em motores elétricos.

### Pergunta de negócio

> Como identificar antecipadamente comportamentos associados a falhas dos motores?

---

## 🔹 2. Entendimento dos dados

Precisamos descobrir:

- quais dados existem;
- como foram coletados;
- quais variáveis estão disponíveis;
- quantidade de registros;
- qualidade dos dados;
- limitações.

---

## 🔹 3. Preparação dos dados

É uma das etapas que normalmente consome bastante tempo.

Inclui:

- limpeza;
- tratamento de valores ausentes;
- tratamento de outliers;
- transformação;
- seleção de atributos;
- criação de novas variáveis.

---

## 🔹 4. Modelagem

Aplicamos algoritmos.

Exemplos:

- classificação;
- regressão;
- clusterização;
- associação;
- redes neurais.

---

## 🔹 5. Avaliação

Perguntamos:

- O modelo funciona?
- O resultado é confiável?
- O resultado resolve o problema?
- Existe algum viés?
- O resultado pode ser utilizado na prática?

---

## 🔹 6. Implantação

O conhecimento produzido precisa chegar ao usuário.

Pode ser por meio de:

- dashboard;
- API;
- sistema web;
- aplicativo;
- relatório;
- sistema de alertas;
- integração com outros sistemas.


# 🔁 4. KDD × CRISP-DM

Os dois processos possuem objetivos semelhantes, mas não são exatamente iguais.

| KDD | CRISP-DM |
|---|---|
| Seleção | Entendimento dos dados |
| Pré-processamento | Preparação dos dados |
| Transformação | Preparação dos dados |
| Mineração | Modelagem |
| Interpretação | Avaliação |
| Conhecimento | Implantação/resultado |

O **KDD** enfatiza a descoberta de conhecimento.

O **CRISP-DM** apresenta uma visão mais ampla do projeto, começando pelo problema de negócio e chegando à implantação.

Para nossos projetos da disciplina, utilizaremos o **CRISP-DM como referência de organização**.

# 💻 7. Relembrando nosso DataFrame

Vamos recriar a base utilizada na Aula 1.


````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

dados = {
    "motor": ["M001", "M001", "M001", "M001", "M001",
              "M002", "M002", "M002", "M002", "M002"],
    "data_hora": [
        "2026-01-01 08:00",
        "2026-01-01 09:00",
        "2026-01-01 10:00",
        "2026-01-01 11:00",
        "2026-01-01 12:00",
        "2026-01-01 08:00",
        "2026-01-01 09:00",
        "2026-01-01 10:00",
        "2026-01-01 11:00",
        "2026-01-01 12:00"
    ],
    "corrente": [12.4, 12.8, 13.7, 15.2, 15.8, 10.8, 11.0, 11.2, 11.5, 11.7],
    "tensao": [380, 379, 381, 378, 377, 380, 381, 380, 379, 380],
    "vibracao": [1.8, 2.1, 3.4, 5.8, 6.2, 1.2, 1.3, 1.4, 1.5, 1.6],
    "temperatura": [62.3, 64.1, 68.7, 76.4, 79.2, 55.1, 56.0, 56.8, 57.4, 58.2],
    "rpm": [1750, 1748, 1742, 1728, 1719, 1752, 1750, 1749, 1748, 1747],
    "potencia": [7.2, 7.4, 8.1, 9.2, 9.6, 6.1, 6.2, 6.3, 6.4, 6.5],
    "horas_operacao": [1250, 1251, 1252, 1253, 1254, 820, 821, 822, 823, 824],
    "status": ["Normal", "Normal", "Alerta", "Falha", "Falha",
               "Normal", "Normal", "Normal", "Normal", "Normal"]
}

df = pd.DataFrame(dados)
df
````

# 📊 9. Primeiras análises

Agora vamos comparar algumas variáveis.

Primeiro, vamos calcular a temperatura média por status.


````python
df.groupby("status")["temperatura"].mean()
````

Observe o resultado.

Pergunta:

> **Os motores classificados como "Falha" apresentam temperatura média diferente dos motores "Normais"?**

Neste momento estamos apenas observando os dados. Ainda não podemos afirmar que temperatura elevada **causa** a falha.


````python
df.groupby("status")["vibracao"].mean()
````

Agora observe a vibração.

Pergunta:

> **O comportamento da vibração parece diferenciar os estados do motor?**


````python
df.groupby("status")[["corrente", "temperatura", "vibracao", "rpm", "potencia"]].mean()
````

Esse tipo de tabela já começa a transformar os dados em informação.

Estamos comparando características dos motores em diferentes estados.

Mais adiante aprenderemos técnicas estatísticas e algoritmos para investigar essas relações de maneira mais rigorosa.


# 📈 10. Visualizando uma possível relação

Vamos criar um gráfico simples relacionando temperatura e vibração.


````python
plt.figure(figsize=(8, 5))

for status in df["status"].unique():
    dados_status = df[df["status"] == status]
    plt.scatter(
        dados_status["temperatura"],
        dados_status["vibracao"],
        label=status
    )

plt.xlabel("Temperatura (°C)")
plt.ylabel("Vibração (mm/s)")
plt.title("Temperatura × Vibração")
plt.legend()
plt.show()
````

### Interpretação

Observe a distribuição dos pontos.

Pergunte:

- Os registros normais estão concentrados em determinada região?
- Os registros de falha aparecem em outra região?
- Existe uma tendência?
- Temos dados suficientes para tirar uma conclusão definitiva?

> **Importante:** um gráfico pode sugerir uma hipótese, mas não necessariamente comprova uma relação causal.
