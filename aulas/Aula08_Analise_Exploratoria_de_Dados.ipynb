# 🧠 1. O que é Análise Exploratória de Dados?

Antes de aplicar um algoritmo de Mineração de Dados, precisamos conhecer os dados.

A **EDA — Exploratory Data Analysis** é o processo de investigar uma base para descobrir:

- padrões;
- tendências;
- distribuições;
- relações;
- anomalias;
- valores ausentes;
- possíveis erros;
- variáveis relevantes.

Podemos pensar:

```text
Dados
  ↓
Conhecer os dados
  ↓
EDA
  ↓
Encontrar padrões
  ↓
Formular hipóteses
  ↓
Preparar os dados
  ↓
Mineração de Dados
```

> **Regra importante:** não devemos aplicar um algoritmo simplesmente porque sabemos executar o algoritmo. Primeiro precisamos entender o problema e os dados.

# 📥 4. Criando nossa base de dados

Vamos criar uma base maior para permitir uma análise mais interessante.

Cada registro representa uma leitura de um motor.

Para facilitar a aula, vamos gerar dados simulados com características semelhantes às encontradas em sensores industriais.


````python
np.random.seed(42)

n = 300

df = pd.DataFrame({
    "motor": np.random.choice(["M001", "M002", "M003", "M004", "M005"], n),
    "temperatura": np.random.normal(65, 8, n),
    "vibracao": np.random.normal(2.2, 0.7, n),
    "corrente": np.random.normal(13, 2, n),
    "tensao": np.random.normal(380, 4, n),
    "rpm": np.random.normal(1740, 15, n)
})

df.head()
````

# 📌 8. Média, mediana e moda

A média é conhecida, mas não é suficiente.

````python
media = df["temperatura"].mean()
mediana = df["temperatura"].median()

print("Média:", media)
print("Mediana:", mediana)
````

A mediana divide os dados em duas partes.

# 📈 9. Distribuição da temperatura

Vamos visualizar como as temperaturas estão distribuídas.

# 📦 10. Boxplot

Outra visualização muito importante é o **boxplot**.



````python
plt.figure(figsize=(8, 4))
plt.boxplot(df["temperatura"], vert=False)
plt.title("Boxplot da Temperatura")
plt.xlabel("Temperatura (°C)")
plt.show()
````

O boxplot permite observar:

# ⚠️ 11. Investigando valores extremos

Vamos localizar registros com temperatura acima de 85 °C.


````python
df[df["temperatura"] > 85].sort_values("temperatura", ascending=False).head(10)
````

Agora vamos contar quantos registros ultrapassam esse limite.


````python
quantidade = (df["temperatura"] > 85).sum()

print("Quantidade de registros:", quantidade)
````

# 🔬 12. Comparando motores

Queremos descobrir se alguns motores trabalham em temperaturas maiores.



````python
temperatura_motor = (
    df.groupby("motor")["temperatura"]
      .agg(["count", "mean", "min", "max"])
      .sort_values("mean", ascending=False)
)

temperatura_motor
````

Isso já nos permite formular uma hipótese:

> **Alguns motores apresentam comportamento térmico diferente dos demais.**

Mas precisamos investigar antes de concluir.


# 📊 13. Gráfico por motor



````python
media_motor = df.groupby("motor")["temperatura"].mean().sort_values()

plt.figure(figsize=(8, 4))
plt.bar(media_motor.index, media_motor.values)
plt.title("Temperatura Média por Motor")
plt.xlabel("Motor")
plt.ylabel("Temperatura média (°C)")
plt.show()
````

Visualizações podem transformar uma tabela difícil de interpretar em uma informação muito mais clara.


# 🔗 14. Relação entre duas variáveis

Agora queremos investigar:

> **A vibração aumenta quando a temperatura aumenta?**

Para isso podemos utilizar um gráfico de dispersão.


````python
plt.figure(figsize=(7, 5))
plt.scatter(df["vibracao"], df["temperatura"], alpha=0.6)
plt.title("Temperatura × Vibração")
plt.xlabel("Vibração")
plt.ylabel("Temperatura (°C)")
plt.show()
````

O gráfico de dispersão ajuda a identificar possíveis relações entre variáveis.

Observe:

```text
X → vibração
Y → temperatura
```

Se os pontos apresentarem uma tendência, podemos investigar a relação com mais profundidade.


# 📐 15. Correlação

A correlação mede o grau de associação linear entre duas variáveis.

Vamos calcular a matriz de correlação.


````python
correlacao = df.select_dtypes(include="number").corr()

correlacao
````

A interpretação simplificada:

```text
+1 → relação positiva forte
 0 → pouca ou nenhuma relação linear
-1 → relação negativa forte
```

Mas atenção:

> **Correlação não significa causalidade.**

Se duas variáveis apresentam correlação, isso não prova que uma causa a outra.


# 🔥 16. Visualizando a correlação

Vamos visualizar a matriz utilizando `imshow()`.


````python
corr = df.select_dtypes(include="number").corr()

plt.figure(figsize=(8, 6))
plt.imshow(corr, interpolation="nearest")
plt.xticks(range(len(corr.columns)), corr.columns, rotation=45, ha="right")
plt.yticks(range(len(corr.columns)), corr.columns)
plt.title("Matriz de Correlação")
plt.colorbar()
plt.tight_layout()
plt.show()
````

# 🧹 17. Verificando valores ausentes

Antes de avançar, precisamos descobrir se existem valores ausentes.



````python
df.isnull().sum()
````

Uma base com valores ausentes pode exigir tratamento antes da aplicação de algoritmos.

Possíveis estratégias:

- remover registros;
- preencher com média;
- preencher com mediana;
- utilizar valor mais frequente;
- utilizar técnicas específicas de imputação.

A escolha depende do problema.


# 🧪 18. Criando uma situação com dados ausentes

Vamos criar alguns valores ausentes propositalmente.


````python
df_eda = df.copy()

df_eda.loc[[5, 20, 50], "temperatura"] = np.nan
df_eda.loc[[10, 30], "vibracao"] = np.nan

df_eda.isnull().sum()
````

Agora podemos comparar a média antes e depois do tratamento.


````python
print("Média com NaN:", df_eda["temperatura"].mean())

df_eda["temperatura"] = df_eda["temperatura"].fillna(
    df_eda["temperatura"].median()
)

print("Média após preenchimento:", df_eda["temperatura"].mean())
````

# 🧠 19. Criando indicadores

EDA também pode ajudar a criar novas variáveis.

Por exemplo, podemos criar uma indicação simples de alerta térmico:

```text
temperatura >= 80 → alerta
temperatura < 80 → normal
```


````python
df["alerta_temperatura"] = np.where(
    df["temperatura"] >= 80,
    "Alerta",
    "Normal"
)

df[["motor", "temperatura", "alerta_temperatura"]].head(10)
````

Agora podemos contar quantos alertas existem.


````python
df["alerta_temperatura"].value_counts()
````

# 🔎 20. Investigando múltiplas variáveis

Vamos criar um indicador simples combinando temperatura e vibração.

> Esse indicador é apenas didático. Em um sistema real, os limites deveriam ser definidos com base em especificações técnicas e conhecimento do domínio.


````python
df["nivel_atencao"] = np.where(
    (df["temperatura"] >= 80) & (df["vibracao"] >= 3),
    "Alto",
    "Normal"
)

df["nivel_atencao"].value_counts()
````

Agora vamos investigar quais motores apresentam mais situações de atenção.


````python
pd.crosstab(df["motor"], df["nivel_atencao"])
````
