````python
import pandas as pd

# No projeto:
# df = pd.read_csv("dados.csv")

# Exemplo demonstrativo
df = pd.DataFrame({
    "categoria": ["A", "B", "A", "C", "B"],
    "valor": [100, 250, 150, 400, 300],
    "quantidade": [2, 5, 3, 8, 6]
})

df.head()
````

# 🔎 8. Primeira inspeção

````python
print("Dimensões:", df.shape)

print("\nColunas:")
print(df.columns.tolist())

print("\nTipos:")
print(df.dtypes)

print("\nValores ausentes:")
print(df.isnull().sum())

print("\nDuplicados:", df.duplicated().sum())
````

# 📈 10. Visualização

Exemplo de distribuição:


````python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))
plt.hist(df["valor"], bins=10)
plt.title("Distribuição dos valores")
plt.xlabel("Valor")
plt.ylabel("Quantidade")
plt.show()
````

Exemplo de relação entre duas variáveis:


````python
plt.figure(figsize=(8, 5))
plt.scatter(df["quantidade"], df["valor"])
plt.title("Quantidade × Valor")
plt.xlabel("Quantidade")
plt.ylabel("Valor")
plt.show()
````
