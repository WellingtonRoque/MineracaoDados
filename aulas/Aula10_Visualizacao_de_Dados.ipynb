# 🧠 1. Por que visualizar os dados?

Uma tabela pode conter centenas ou milhares de registros.

Por exemplo:

| Motor | Temperatura | Vibração | Corrente |
|---|---:|---:|---:|
| M001 | 62.4 | 1.8 | 12.3 |
| M002 | 65.2 | 2.1 | 13.1 |
| M003 | 78.4 | 3.2 | 15.4 |
| ... | ... | ... | ... |

É possível analisar a tabela, mas um gráfico pode revelar rapidamente:

- tendências;
- diferenças;
- concentração;
- anomalias;
- relações entre variáveis.

```text
DADOS
  ↓
VISUALIZAÇÃO
  ↓
PADRÕES
  ↓
INTERPRETAÇÃO
  ↓
DECISÃO
```

> Um gráfico não substitui a análise. Ele é uma ferramenta para ajudar a enxergar os dados.

# 💻 3. Preparando o ambiente

# 📥 4. Criando a base de trabalho

Vamos gerar uma base simulada de leituras industriais.



````python
n = 300

df = pd.DataFrame({
    "motor": np.random.choice(["M001", "M002", "M003", "M004", "M005"], n),
    "temperatura": np.random.normal(65, 8, n),
    "vibracao": np.random.normal(2.2, 0.7, n),
    "corrente": np.random.normal(13, 2, n),
    "tensao": np.random.normal(380, 4, n),
    "rpm": np.random.normal(1740, 15, n)
})

df.head()
````

# 📊 6. Gráfico de barras

Queremos comparar a temperatura média dos motores.



````python
media_motor = (
    df.groupby("motor")["temperatura"]
      .mean()
      .sort_values(ascending=False)
)

plt.figure(figsize=(8, 5))
plt.bar(media_motor.index, media_motor.values)
plt.title("Temperatura Média por Motor")
plt.xlabel("Motor")
plt.ylabel("Temperatura média (°C)")
plt.show()
````

# 📈 7. Gráfico de linha

O gráfico de linha é adequado para observar **evolução ao longo do tempo**.

Vamos criar uma sequência temporal para um motor.


````python
datas = pd.date_range(
    start="2026-03-01 08:00",
    periods=48,
    freq="h"
)

temperaturas = 65 + np.linspace(0, 18, 48) + np.random.normal(0, 2, 48)

df_tempo = pd.DataFrame({
    "data_hora": datas,
    "temperatura": temperaturas
})

df_tempo.head()
````

````python
plt.figure(figsize=(10, 5))
plt.plot(df_tempo["data_hora"], df_tempo["temperatura"])
plt.title("Evolução da Temperatura")
plt.xlabel("Data e hora")
plt.ylabel("Temperatura (°C)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

Observe se existe:

- tendência de aumento;
- tendência de redução;
- estabilidade;
- picos;
- mudanças bruscas.

Essa análise pode ser importante para manutenção preditiva.


# 📉 8. Criando uma situação de alerta

Vamos destacar mentalmente a ideia de limite.

Suponha que a equipe de manutenção considere:

```text
Temperatura < 80 °C → operação normal
Temperatura ≥ 80 °C → atenção
```

Podemos visualizar esse limite junto com os dados.


````python
plt.figure(figsize=(10, 5))
plt.plot(df_tempo["data_hora"], df_tempo["temperatura"], label="Temperatura")
plt.axhline(80, linestyle="--", label="Limite de atenção")
plt.title("Temperatura e Limite de Atenção")
plt.xlabel("Data e hora")
plt.ylabel("Temperatura (°C)")
plt.legend()
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
````

# 🔵 11. Gráfico de dispersão

Agora queremos investigar uma relação:

> **Temperatura × Vibração**


````python
plt.figure(figsize=(8, 5))
plt.scatter(
    df["vibracao"],
    df["temperatura"],
    alpha=0.6
)
plt.title("Temperatura × Vibração")
plt.xlabel("Vibração")
plt.ylabel("Temperatura (°C)")
plt.show()
````

Observe a posição dos pontos.

Se houver uma tendência crescente, podemos investigar uma possível associação entre as variáveis.

Mas lembre:

> **Visualizar uma relação não significa provar causalidade.**


# 🎯 12. Colorindo categorias

Podemos utilizar uma variável categórica para facilitar a comparação.

Vamos criar um indicador simples.


````python
df["situacao"] = np.where(
    (df["temperatura"] >= 80) | (df["vibracao"] >= 3),
    "Atenção",
    "Normal"
)

df["situacao"].value_counts()
````

Agora vamos separar visualmente as situações.


````python
for situacao in df["situacao"].unique():
    dados = df[df["situacao"] == situacao]
    plt.scatter(
        dados["vibracao"],
        dados["temperatura"],
        alpha=0.6,
        label=situacao
    )

plt.title("Temperatura × Vibração por Situação")
plt.xlabel("Vibração")
plt.ylabel("Temperatura (°C)")
plt.legend()
plt.show()
````

Essa técnica é muito útil para observar se determinada região do gráfico concentra situações de interesse.

# ⚠️ 16. Erros comuns em visualização

## 1. Gráfico sem título

O leitor não sabe o que está vendo.

## 2. Eixos sem identificação

Não fica claro o que cada eixo representa.

## 3. Excesso de informação

Muitos elementos dificultam a leitura.

## 4. Gráfico inadequado

Um gráfico de pizza, por exemplo, não é adequado para qualquer comparação.

## 5. Não interpretar o gráfico

Mostrar o gráfico não é apresentar a análise.

## 6. Manipular a escala

Uma escala inadequada pode exagerar ou esconder diferenças.
