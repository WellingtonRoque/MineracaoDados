# 3. 💎 Missão 1 — Cristal mágico

Vamos começar com o cristal.

Queremos que ele alterne continuamente entre azul e roxo.

### Preparando o Script

1. No Explorer, localize a peça `Cristal`.
2. Clique com o botão direito sobre ela.
3. Selecione **Inserir objeto → Script**.
4. Apague o conteúdo inicial.
5. Digite o código abaixo.

### Código — Cristal mágico

```lua
while true do

    script.Parent.Color = Color3.fromRGB(0, 170, 255)

    task.wait(1)

    script.Parent.Color = Color3.fromRGB(170, 0, 255)

    task.wait(1)

end
```

### Execute o jogo

Observe o cristal:

- primeiro fica azul;
- espera aproximadamente 1 segundo;
- depois fica roxo;
- espera aproximadamente 1 segundo;
- volta ao azul.

E esse comportamento continua durante o jogo.


# 4. 🔍 Entendendo o Script do cristal

Vamos analisar as partes mais importantes.

### A repetição

```lua
while true do
```

Significa que a repetição continuará, pois `true` é sempre verdadeiro.

### Alterando a cor

```lua
script.Parent.Color = Color3.fromRGB(0, 170, 255)
```

`script.Parent` representa a peça onde o Script está inserido.

A propriedade `Color` altera a cor dessa peça.

### A pausa

```lua
task.wait(1)
```

O Script aguarda aproximadamente um segundo antes de executar a próxima instrução.

### Encerrando o bloco

```lua
end
```

Fecha o bloco do `while`.

Depois do `end`, como a condição é `true`, o laço começa novamente.

# 7. 🚨 Missão 3 — Sinal de perigo

Até agora utilizamos `while true`, que mantém a repetição acontecendo continuamente.

Agora vamos criar um comportamento diferente.

O sinal de perigo deverá piscar **cinco vezes** e depois parar.

Para isso, vamos utilizar um contador e uma condição no `while`.

### Preparação

1. Selecione a peça `SinalPerigo`.
2. Insira um `Script` dentro dela.
3. Apague o código inicial.

### Código — Sinal de perigo

```lua
local contador = 1

while contador <= 5 do

    script.Parent.Color = Color3.fromRGB(255, 0, 0)

    task.wait(0.5)

    script.Parent.Color = Color3.fromRGB(255, 255, 0)

    task.wait(0.5)

    contador = contador + 1

end

script.Parent.Color = Color3.fromRGB(80, 80, 80)
```

Execute o jogo e observe.


# 8. 🔍 Entendendo o sinal de perigo

Observe a primeira linha:

```lua
local contador = 1
```

O contador começa em 1.

Depois temos:

```lua
while contador <= 5 do
```

A repetição continua enquanto o contador for menor ou igual a 5.

Dentro do laço, o sinal:

1. fica vermelho;
2. aguarda meio segundo;
3. fica amarelo;
4. aguarda meio segundo;
5. aumenta o contador em 1.

Quando o contador chega a 6, a condição fica falsa e o `while` termina.

Por último, o código executa:

```lua
script.Parent.Color = Color3.fromRGB(80, 80, 80)
```

O sinal fica cinza.

### O que aprendemos?

Podemos utilizar `while` de duas maneiras:

- `while true do`: repetição contínua;
- `while contador <= 5 do`: repetição controlada por uma condição.
